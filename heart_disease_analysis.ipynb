## Outlier

```python
numeric_cols = [
    "Age",
    "Weight",
    "Height",
    "BMI",
    "Systolic_BP",
    "Diastolic_BP",
    "Heart_Rate",
    "Blood_Sugar_Fasting",
    "Cholesterol_Total"
]

plt.figure(figsize=(15, 8))

sns.boxplot(data=df[numeric_cols])

plt.xticks(rotation=45)
plt.title("Boxplot of Numerical Variables")
plt.ylabel("Value")

plt.show()
```

```python
outlier_summary = []

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ]

    outlier_summary.append({
        "Column": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": len(outliers),
        "Outlier %": len(outliers) / len(df) * 100
    })

outlier_summary = pd.DataFrame(outlier_summary)

outlier_summary
```

thông qua biểu đồ và bảng dữ liệu, không phát hiện có dữ liệu ngoại lai

# Age Analysis

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="Age",
    bins=20,
    kde=True
)

plt.title("Age Distribution of Patients")
plt.xlabel("Age")
plt.ylabel("Number of Patients")

plt.show()
```

```python
df["Age"].describe()
```

```python
df["Age_Group"] = pd.cut(
    df["Age"],
    bins=[0, 30, 40, 50, 60, 70, 80, 90, float("inf")],
    labels=["<30", "30-39", "40-49", "50-59", "60-69", "70-79", "80-89", "90+"],
    right=False
)

df["Age_Group"].value_counts().sort_index()
```

```python
age_analysis = (
    df.groupby("Age_Group", observed=True)["Heart_Disease"]
      .agg(
          patient_count="count",
          heart_disease_cases="sum",
          heart_disease_rate="mean"
      )
      .reset_index()
)

age_analysis["heart_disease_rate"] *= 100

age_analysis
```

Các bệnh nhân trong nhóm tuổi 60-69, 70-79 có tỷ lệ mắc bệnh cao hơn so với các nhóm tuổi khác.

```python
plt.figure(figsize=(10, 6))

ax = sns.barplot(
    data=age_analysis,
    x="Age_Group",
    y="heart_disease_rate"
)

for container in ax.containers:
    ax.bar_label(
        container,
        fmt="%.1f%%"
    )

plt.title("Heart Disease Rate by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Heart Disease Rate (%)")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=age_analysis,
    x="Age_Group",
    y="heart_disease_cases"
)

plt.title("Heart Disease Cases by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Number of Cases")

plt.show()
```

Tỷ lệ mắc bệnh tim tim tăng dần qua các nhóm tuổi trong bộ dữ liệu này, có nhóm tuổi 60-69 và 70-79 là tỷ lệ cao nhất.

# Clinical Analysis

Phân tích này xem xét các tình trạng lâm sàng, tiền sử bệnh và các chỉ số lâm sàng, đồng thời kiểm tra mối liên hệ của chúng với bệnh tim trong tập dữ liệu này.

```python
clinical_binary_cols = [
    "Hypertension",
    "Diabetes",
    "Hyperlipidemia",
    "Family_History",
    "Previous_Heart_Attack"
]

clinical_analysis = []

for col in clinical_binary_cols:
    temp = (
        df.groupby(col)["Heart_Disease"]
          .agg(
              patient_count="count",
              heart_disease_cases="sum",
              heart_disease_rate="mean"
          )
          .reset_index()
          .rename(columns={col: "Factor_Status"})
    )

    temp["Clinical_Factor"] = col
    temp["heart_disease_rate"] *= 100
    clinical_analysis.append(temp)

clinical_analysis = pd.concat(clinical_analysis, ignore_index=True)
clinical_analysis
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    data=clinical_analysis,
    x="Clinical_Factor",
    y="heart_disease_rate",
    hue="Factor_Status"
)

plt.title("Heart Disease Rate by Clinical Factors")
plt.xlabel("Clinical Factor")
plt.ylabel("Heart Disease Rate (%)")
plt.xticks(rotation=20)

plt.show()
```

# Risk Factor Analysis

```python
risk_factor_cols = [
    "Hypertension",
    "Diabetes",
    "Hyperlipidemia",
    "Family_History",
    "Previous_Heart_Attack"
]

risk_factor_analysis = []

for col in risk_factor_cols:
    temp = (
        df.groupby(col)["Heart_Disease"]
        .agg(
            patient_count="count",
            heart_disease_cases="sum",
            heart_disease_rate="mean"
        )
        .reset_index()
    )

    temp["Risk_Factor"] = col
    temp["heart_disease_rate"] *= 100

    risk_factor_analysis.append(temp)

risk_factor_analysis = pd.concat(
    risk_factor_analysis,
    ignore_index=True
)

risk_factor_analysis
```

```python
risk_factor_summary = []

for col in risk_factor_cols:
    rate_without = (
        df.loc[df[col] == 0, "Heart_Disease"].mean() * 100
    )

    rate_with = (
        df.loc[df[col] == 1, "Heart_Disease"].mean() * 100
    )

    risk_factor_summary.append({
        "Risk Factor": col,
        "Without Factor (%)": rate_without,
        "With Factor (%)": rate_with,
        "Difference (pp)": rate_with - rate_without
    })

risk_factor_summary = pd.DataFrame(risk_factor_summary)

risk_factor_summary
```

# Key Findings

- Trong bộ dữ liệu này, tỷ lệ bệnh tim có xu hướng cao hơn ở nhóm tuổi 60-79.
- Bệnh nhân có tiền sử nhồi máu cơ tim có tỷ lệ bệnh tim cao hơn nhóm không có tiền sử này.
- Tỷ lệ bệnh giữa các nhóm giới tính và một số yếu tố lối sống khá tương đồng.
- Đây là các mối liên hệ trong dữ liệu, không chứng minh nguyên nhân gây bệnh.

# Limitations

- Dữ liệu được sử dụng là dữ liệu tổng hợp, nên kết quả có thể không phản ánh đầy đủ bệnh nhân ngoài thực tế.
- Tỷ lệ dữ liệu thiếu được ghi nhận cao; thay bằng nhãn Unknown có thể ảnh hưởng đến cách diễn giải kết quả.
- Phân tích hiện chủ yếu mô tả và so sánh tỷ lệ, chưa kiểm định ý nghĩa thống kê hoặc kiểm soát nhiều yếu tố cùng lúc.
- Vì vậy, không nên dùng các kết quả này để chẩn đoán hoặc kết luận quan hệ nhân quả.
